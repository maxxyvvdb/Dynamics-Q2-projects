### Derivation of 2a and 2b

$$
\frac{\partial U}{\partial x} + \frac{\partial V}{\partial y} + \frac{\partial W}{\partial z} = 0
$$

Assume geostrophic balance we have:
$$U_g = -\frac{1}{\rho f} \frac{\partial P}{\partial y}$$

$$V_g = \frac{1}{\rho f} \frac{\partial P}{\partial x}$$

Differentiate:
$$
\frac{\partial U_g}{\partial x} = \frac{\partial}{\partial x} \left( -\frac{1}{\rho f} \frac{\partial P}{\partial y} \right) = -\frac{1}{\rho f} \frac{\partial^2 P}{\partial x \partial y}
$$

$$
\frac{\partial V_g}{\partial y} = \frac{\partial}{\partial y} \left( \frac{1}{\rho f} \frac{\partial P}{\partial x} \right) = \frac{1}{\rho f} \frac{\partial^2 P}{\partial y \partial x}
$$

Combine:
$$
\frac{\partial U}{\partial x} + \frac{\partial V}{\partial y} = -\frac{1}{\rho f} \frac{\partial^2 P}{\partial x \partial y} + \frac{1}{\rho f} \frac{\partial^2 P}{\partial y \partial x}
$$

$$
\frac{\partial U}{\partial x} + \frac{\partial V}{\partial y} = 0
$$

$$0 + \frac{\partial W}{\partial z} = 0 \quad \implies \quad \frac{\partial W}{\partial z} = 0$$

from (2):
$$
0 = fV - fV_g - \frac{\partial \overline{u'w'}}{\partial z} \quad \Longrightarrow \quad f(V - V_g) = \frac{\partial \overline{u'w'}}{\partial z} [1]
$$

from (3):
$$
0 = -fU - (-fU_g) - \frac{\partial \overline{v'w'}}{\partial z} \quad \Longrightarrow \quad -f(U - U_g) = \frac{\partial \overline{v'w'}}{\partial z} [2]
$$

Ingegrate [1]:
$$
\int_0^h f(V - V_g) dz = \int_0^h \frac{\partial \overline{u'w'}}{\partial z} dz
$$

$$
f \left( \int_0^h V dz - h V_g \right) = \overline{u'w'}(h) - \overline{u'w'}(sfc) = -\overline{u'w'}(sfc)
$$

$$\int_0^h V dz = h V_g - \frac{1}{f} \overline{u'w'}(sfc) [3]
$$


Integrate [2]:
$$
\int_0^h -f(U - U_g) dz = \int_0^h \frac{\partial \overline{v'w'}}{\partial z} dz
$$
$$
-f \left( \int_0^h U dz - h U_g \right) = \overline{v'w'}(h) - \overline{v'w'}(sfc) = -\overline{v'w'}(sfc)
$$
$$
\int_0^h U dz = h U_g + \frac{1}{f} \overline{v'w'}(sfc) [4]
$$

Integrate (5):
$$
\int_0^h \left( \frac{\partial U}{\partial x} + \frac{\partial V}{\partial y} + \frac{\partial W}{\partial z} \right) dz = 0
$$

$$
\frac{\partial}{\partial x} \left( \int_0^h U dz \right) + \frac{\partial}{\partial y} \left( \int_0^h V dz \right) + (W(h) - W(0)) = 0
$$
at 0, W = 0,
$$
W(h) = - \left[ \frac{\partial}{\partial x} \left( \int_0^h U dz \right) + \frac{\partial}{\partial y} \left( \int_0^h V dz \right) \right]
$$


Substitute [1] and [2] into this:

$$
W(h) = - \left[ \frac{\partial}{\partial x} \left( h U_g + \frac{1}{f} \overline{v'w'}_{sfc} \right) + \frac{\partial}{\partial y} \left( h V_g - \frac{1}{f} \overline{u'w'}_{sfc} \right) \right]
$$

$$
= - \left[ h \left( \frac{\partial U_g}{\partial x} + \frac{\partial V_g}{\partial y} \right) + \frac{1}{f} \frac{\partial \overline{v'w'}_{sfc}}{\partial x} - \frac{1}{f} \frac{\partial \overline{u'w'}_{sfc}}{\partial y} \right]
$$

In 2a we found $\frac{\partial U_g}{\partial x} + \frac{\partial V_g}{\partial y} = 0$, so

$$
W(h) = - \left( \frac{1}{f} \frac{\partial \overline{v'w'}_{sfc}}{\partial x} - \frac{1}{f} \frac{\partial \overline{u'w'}_{sfc}}{\partial y} \right) = \frac{1}{f} \left( \frac{\partial \overline{u'w'}_{sfc}}{\partial y} - \frac{\partial \overline{v'w'}_{sfc}}{\partial x} \right)
$$
